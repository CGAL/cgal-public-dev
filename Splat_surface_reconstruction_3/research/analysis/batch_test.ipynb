```python
import sys
import os

sys.path.append("../")

import utils.poisson_sample as poisson_sampling

meshes_folder = "../../../Data/data/meshes/"
samples_folder = "batch_test_point_clouds/"

meshes_to_run = [
    "pig.off",
    "elephant.off"
]

d = 0.02
```

```python
# make sure all sample files exist for the meshes to run
for mesh_file in meshes_to_run:
    poisson_sampling.verify_samples_file_exists(meshes_folder, samples_folder, mesh_file, d)
```

```python
import os
import subprocess

executable_path = os.path.abspath(
    "../../build/Splat_surface_reconstruction_3/splat_reconstruction_function"
)

print(f"Executable: {executable_path}")

for mesh_file in meshes_to_run:
    mesh_name = os.path.splitext(os.path.basename(mesh_file))[0]
    samples_filename = f"batch_test_point_clouds/{mesh_name}_d{d}.off"

    result = subprocess.run(
        [executable_path, samples_filename],
        capture_output=True,
        text=True
    )

    if result.returncode != 0:
        print(f"Command failed for {samples_filename}")
        print(f"Return code: {result.returncode}")
        print("stdout:")
        print(result.stdout)
        print("stderr:")
        print(result.stderr)
        break
    else:
        print(f"Successfully processed {samples_filename}")
```
